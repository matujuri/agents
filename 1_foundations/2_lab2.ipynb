```python
# Start with imports - ask ChatGPT to explain any package that you don't know

import os
import json
from dotenv import load_dotenv
from openai import OpenAI
from IPython.display import Markdown, display
```

```python
# Always remember to do this!
load_dotenv(override=True)
```

```python
# Print the key prefixes to help with any debugging

openai_api_key = os.getenv('OPENAI_API_KEY')
anthropic_api_key = os.getenv('ANTHROPIC_API_KEY')
google_api_key = os.getenv('GEMINI_API_KEY')
deepseek_api_key = os.getenv('DEEPSEEK_API_KEY')
groq_api_key = os.getenv('GROQ_API_KEY')

if openai_api_key:
    print(f"OpenAI API Key exists and begins {openai_api_key[:8]}")
else:
    print("OpenAI API Key not set")
    
if anthropic_api_key:
    print(f"Anthropic API Key exists and begins {anthropic_api_key[:7]}")
else:
    print("Anthropic API Key not set (and this is optional)")

if google_api_key:
    print(f"Google API Key exists and begins {google_api_key[:2]}")
else:
    print("Google API Key not set (and this is optional)")

if deepseek_api_key:
    print(f"DeepSeek API Key exists and begins {deepseek_api_key[:3]}")
else:
    print("DeepSeek API Key not set (and this is optional)")

if groq_api_key:
    print(f"Groq API Key exists and begins {groq_api_key[:4]}")
else:
    print("Groq API Key not set (and this is optional)")
```

```python
question = "AIとユーモアのビジネスチャンスを幅広く考えて、一つのみ提案ください。ビジネスチャンスのみでお願いします。"
```

```python
competitors = []
answers = []
```

```python
model_name = "gpt-5-nano"
openai = OpenAI()
response = openai.responses.create(
    model=model_name,
    input=question
)
answer_openai = response.output_text
display(Markdown(answer_openai))
competitors.append(model_name)
answers.append(answer_openai)
```

```python
# Anthropic has a slightly different API, and Max Tokens is required
from anthropic import Anthropic
anthropic_api_key = os.getenv('ANTHROPIC_API_KEY')
model_name = "claude-sonnet-4-20250514"

client = Anthropic(
    # defaults to os.environ.get("ANTHROPIC_API_KEY")
)
message = client.messages.create(
    model=model_name,
    max_tokens=1024,
    messages=[
        {"role": "user", "content": question}
    ]
)
answer_anthropic = message.content[0].text

display(Markdown(answer_anthropic))
competitors.append(model_name)
answers.append(answer_anthropic)
```

```python
model_name = "gemini-2.5-flash"

from google import genai
# The client gets the API key from the environment variable `GEMINI_API_KEY`.
client = genai.Client()
response = client.models.generate_content(
    model=model_name, contents=question
)
answer_gemini = response.text
display(Markdown(answer_gemini))
competitors.append(model_name)
answers.append(answer_gemini)
```

```python
messages = [{"role": "user", "content": question}]
deepseek_api_key = os.getenv('DEEPSEEK_API_KEY')
deepseek = OpenAI(api_key=deepseek_api_key, base_url="https://api.deepseek.com")
model_name = "deepseek-chat"

response = deepseek.chat.completions.create(model=model_name, messages=messages)
answer_deepseek = response.choices[0].message.content

display(Markdown(answer_deepseek))
competitors.append(model_name)
answers.append(answer_deepseek)
```

```python
from groq import Groq
groq_api_key = os.getenv('GROQ_API_KEY')
messages = [{"role": "user", "content": question}]
model_name = "llama-3.3-70b-versatile"
client = Groq(
    api_key=groq_api_key,
)

chat_completion = client.chat.completions.create(
    messages=messages,
    model=model_name,
)
answer_groq = chat_completion.choices[0].message.content

display(Markdown(answer_groq))
competitors.append(model_name)
answers.append(answer_groq)
```

## For the next cell, we will use Ollama

Ollama runs a local web service that gives an OpenAI compatible endpoint,  
and runs models locally using high performance C++ code.

If you don't have Ollama, install it here by visiting https://ollama.com then pressing Download and following the instructions.

After it's installed, you should be able to visit here: http://localhost:11434 and see the message "Ollama is running"

You might need to restart Cursor (and maybe reboot). Then open a Terminal (control+\`) and run `ollama serve`

Useful Ollama commands (run these in the terminal, or with an exclamation mark in this notebook):

`ollama pull <model_name>` downloads a model locally  
`ollama ls` lists all the models you've downloaded  
`ollama rm <model_name>` deletes the specified model from your downloads

```python
model_name = "gemma3"

messages = [{"role": "user", "content": f"{question} 250文字以内で回答してください。"}]
from ollama import chat
from ollama import ChatResponse

response: ChatResponse = chat(model=model_name, messages=messages)
answer_ollama = response.message.content

display(Markdown(answer_ollama))
competitors.append(model_name)
answers.append(answer_ollama)
```

```python
model_name = "deepseek-r1:1.5b"

messages = [{"role": "user", "content": f"{question} 250文字以内で回答してください。"}]

response: ChatResponse = chat(model=model_name, messages=messages)
answer_ollama = response.message.content

display(Markdown(answer_ollama))
competitors.append(model_name)
answers.append(answer_ollama)
```

```python
# So where are we?

print(competitors)
print(answers)
```

```python
# It's nice to know how to use "zip"
for competitor, answer in zip(competitors, answers):
    print(f"Competitor: {competitor}\n\n{answer}")
```

```python
# Let's bring this together - note the use of "enumerate"

together = ""
for index, answer in enumerate(answers):
    together += f"# Response from competitor {index+1}\n\n"
    together += answer + "\n\n"
```

```python
print(together)
```

```python
judge = f"""あなたは、{len(competitors)} 人の競技者の間で行われる「ビジネスチャンス提案コンペティション」の審査を担当しています。  
各モデルには次の質問が与えられています：

{question}

あなたの役割は、それぞれの提案を以下の基準で評価し、最も優れたものから順にランク付けすることです：

1. 明確さ – 提案内容がわかりやすく伝わるか  
2. 独自性 – アイデアに新規性や差別化があるか  
3. 実現可能性 – 技術的・経済的に実行できそうか  
4. 市場性 – 潜在的な需要や顧客層が見込めるか  
5. 成長性 – 将来的に拡大・発展の余地があるか  
6. 社会的価値 – 社会やユーザーにポジティブな影響を与えるか  

JSON形式で、以下の形式でのみ回答してください：
{{"results": ["最も優れた競技者の番号", "次に優れた競技者の番号", "3番目に優れた競技者の番号", ...]}}

各競技者からの回答は以下の通りです：

{together}

ここで、競技者のランク順を示すJSONだけを回答してください。他の内容は含めないでください。Markdown形式やコードブロックも使用しないでください。"""
```

```python
print(judge)
```

```python
# Judgement time!

model_name = "gpt-5-nano"
openai = OpenAI()
response = openai.responses.create(
    model=model_name,
    input=judge
)
results = response.output_text
print(results)
```

```python
# OK let's turn this into results!

results_dict = json.loads(results)
ranks = results_dict["results"]
for index, result in enumerate(ranks):
    competitor = competitors[int(result)-1]
    print(f"Rank {index+1}: {competitor}")
```
